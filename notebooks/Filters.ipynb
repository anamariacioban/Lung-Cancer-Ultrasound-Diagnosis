```python
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

cale_imagine = r'/Users/ana/Downloads/Lung Cancer Diagnosis System/Imagini pentru analiza/cancere/tu pulm perif 1.jpg'  

try:
    img_pil = Image.open(cale_imagine)
    img_gray = np.array(img_pil.convert('L')) 

   
    img_bilateral = cv2.bilateralFilter(img_gray, 9, 75, 75)
    img_final = cv2.equalizeHist(img_bilateral)
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(img_gray, cmap='gray')
    plt.title('Originală (Grayscale)')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(img_bilateral, cmap='gray')
    plt.title('După Filtru Bilateral')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.imshow(img_final, cmap='gray')
    plt.title('Bilateral + Egalizare Histogramă')
    plt.axis('off')
    
    plt.show()

except FileNotFoundError:
    print("Eroare: Calea incorectă. Verifică din nou calea completă a fișierului!")
except Exception as e:
    print(f"A apărut o eroare neașteptată: {e}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps 

cale_imagine =r'/Users/ana/Downloads/Lung Cancer Diagnosis System/Imagini pentru analiza/cancere/tu pulm perif 1.jpg'  

OFFSET_DECUPEAZA_SUS = 10 
OFFSET_DECUPEAZA_JOS = 10 

try:
    img_pil = Image.open(cale_imagine)
    img_gray = img_pil.convert('L') 
    img_np_contrast = np.array(ImageOps.equalize(img_gray))
    height, width = img_np_contrast.shape

    _, img_binara_limite = cv2.threshold(img_np_contrast, 50, 255, cv2.THRESH_BINARY)
    contururi, _ = cv2.findContours(img_binara_limite, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    contur_mare = max(contururi, key=cv2.contourArea)
    x, y, w, h = cv2.boundingRect(contur_mare)
    
    start_x_decupat = x
    end_x_decupat = x + w
    start_y_decupat = y + OFFSET_DECUPEAZA_SUS
    end_y_decupat = (y + h) - OFFSET_DECUPEAZA_JOS 
    start_y_decupat = max(0, start_y_decupat)
    end_y_decupat = min(height, end_y_decupat)
    
    
    img_decupata = img_np_contrast[start_y_decupat:end_y_decupat, start_x_decupat:end_x_decupat]
    
    prag_ostu_decupata, _ = cv2.threshold(img_decupata, 0, 255, cv2.THRESH_BINARY | cv2.THRESH_OTSU)
    _, img_izolata = cv2.threshold(img_decupata, prag_ostu_decupata, 255, cv2.THRESH_BINARY_INV)


    #numărăm pixelii albi din imaginea izolată
    numar_pixeli_patologie = np.sum(img_izolata == 255)
    suprafata_totala_decupata = img_izolata.size
    procent_patologie = (numar_pixeli_patologie / suprafata_totala_decupata) * 100
    
    contururi_patologie, _ = cv2.findContours(img_izolata, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    # convertim imaginea decupată în BGR pentru a putea desena contururi colorate
    img_marcata = cv2.cvtColor(img_decupata, cv2.COLOR_GRAY2BGR) 

    # desenăm contururile găsite 
    cv2.drawContours(img_marcata, contururi_patologie, -1, (0, 255, 0), 2)
    
    # 4. afisare
    plt.figure(figsize=(18, 6))
    
    # imagine1
    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img_marcata, cv2.COLOR_BGR2RGB))
    plt.title(f'1. Patologie Marcată pe Decupaj ({procent_patologie:.2f}% din zonă)')
    plt.axis('off')
    
    # im2
    plt.subplot(1, 3, 2)
    plt.imshow(img_izolata, cmap='gray')
    plt.title(f'2. Patologie Izolată (Otsu={int(prag_ostu_decupata)})')
    plt.axis('off')

    # Histograma 
    plt.subplot(1, 3, 3)
    plt.hist(img_decupata.ravel(), bins=256, range=[0, 256])
    plt.axvline(x=prag_ostu_decupata, color='r', linestyle='--', linewidth=1, label=f'Prag Otsu: {int(prag_ostu_decupata)}')
    plt.title('3. Histograma (Ajustare Prag)')
    plt.legend()
    
    plt.show()
    
    print("-" * 50)
    print(f"REZULTAT DIAGNOSTIC AUTOMAT:")
    print(f"Suprafața decupată: {suprafata_totala_decupata} pixeli")
    print(f"Pixeli Patologici Izolați: {numar_pixeli_patologie}")
    print(f"Procent Patologie din Zona de Interes: {procent_patologie:.2f}%")
    print("-" * 50)


except Exception as e:
    print(f"A apărut o eroare: {e}")
```
