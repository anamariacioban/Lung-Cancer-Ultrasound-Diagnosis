```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# CONFIGURARE ȘI CALE
cale_imagine = r'/Users/ana/Downloads/Lung Cancer Diagnosis System/Imagini pentru analiza/cancere/tu pulm perif 1.jpg' 

try:
    img_pil = Image.open(cale_imagine)
    img_bgr = np.array(img_pil)
    img_color_original = cv2.cvtColor(img_bgr, cv2.COLOR_RGB2BGR)

    # DELIMITARE ECOGRAFIE (CONTUR ALB CURBAT ȘI ÎNCHIS)

    img_cu_delimitare_alba = img_color_original.copy()
    img_gray_delimitare = cv2.cvtColor(img_color_original, cv2.COLOR_BGR2GRAY)

    # Egalizare Histogramei (CLAHE) - Uniformizează imaginea
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    img_np_contrast = clahe.apply(img_gray_delimitare)

    # Netezire Gaussiană fină 
    img_blurred = cv2.GaussianBlur(img_np_contrast, (9, 9), 0)
    
    # Prag (Otsu) automat 
    _, thresh_delimitare = cv2.threshold(img_blurred, 0, 255, cv2.THRESH_BINARY | cv2.THRESH_OTSU)
    
    # Ultra-Netezire Morfologică
    kernel_morph = np.ones((10, 10), np.uint8) # Kernel mărit
    thresh_clean = cv2.morphologyEx(thresh_delimitare, cv2.MORPH_CLOSE, kernel_morph)

    # Găsirea contururilor
    contours_delimitare, _ = cv2.findContours(thresh_clean, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    
    if contours_delimitare:
        largest_contour_delimitare = max(contours_delimitare, key=cv2.contourArea)
        
        # Desenează conturul ALB
        cv2.drawContours(img_cu_delimitare_alba, [largest_contour_delimitare], -1, (255, 255, 255), 2)
    else:
        print(" Nu s-a putut trasa conturul alb de delimitare.")
        
    # VIZUALIZARE REZULTAT
    
    plt.figure(figsize=(8, 8))
    plt.imshow(cv2.cvtColor(img_cu_delimitare_alba, cv2.COLOR_BGR2RGB))
    plt.title("Contur Alb Final (Arcuit și Închis)")
    plt.axis('off')
    plt.show()

    print("\n Conturul alb a fost trasat ca o figură închisă, cu arce și linii laterale cât mai netede, folosind CLAHE și Otsu adaptiv.")
    
except Exception as e:
    print(f"A apărut o eroare: {e}")
```
